```python
%sql

create catalog if not exists zeyo_catalog;

create schema if not exists zeyo_schema;

create volume if not exists zeyo_volume;

INSERT OVERWRITE DIRECTORY
'/Volumes/zeyo_catalog/zeyo_schema/zeyo_volume/zeyodata/'
USING TEXT
VALUES
  ('1,sai'),
  ('2,zeyo');
```

```python
%sql
select 
  split(value, ',')[0] as id,
  split(value, ',')[1] as name 
from text.`/Volumes/zeyo_catalog/zeyo_schema/zeyo_volume/zeyodata`;
```

```python
%sql
create view zeyo_view as 
select 
  split(value, ',')[0] as id,
  split(value, ',')[1] as name 
from text.`/Volumes/zeyo_catalog/zeyo_schema/zeyo_volume/zeyodata`;
```

```python
%sql
select * from zeyo_view;
```
