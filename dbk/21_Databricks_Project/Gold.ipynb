```python
%sql
create or replace table procat.procat_schema.gold_orders (
    order_id STRING,
    order_date STRING,
    customer_id STRING,
    region STRING,
    product_name STRING,
    unit_price DECIMAL(12,2),
    quantity INT,
    order_amount DECIMAL(14,2),
    order_year INT,
    order_month INT
)
USING DELTA
PARTITIONED BY (order_year, order_month)
--LOCATION 's3://zeyob48/targetorder/finalGoldd/';
```

```python
%sql
INSERT OVERWRITE TABLE procat.procat_schema.gold_orders
PARTITION (order_year, order_month)

SELECT 
    order_id,
    order_date,
    customer_id,
    region,
    product_name,
    unit_price,
    quantity,
    order_amt,

    YEAR(TO_DATE(order_date)) AS order_year,
    MONTH(TO_DATE(order_date)) AS order_month

FROM procat.procat_schema.sliver_orders;
```
